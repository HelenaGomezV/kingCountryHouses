````python
def categorize_renovation_year_numeric(year):
    if year == 0:
        return 0  # Not Renovated
    elif year < 1980:
        return 1  # Before 1980
    elif year < 1990:
        return 2  # 1980-1989
    elif year < 2000:
        return 3  # 1990-1999
    elif year < 2010:
        return 4  # 2000-2009
    else:
        return 5  # 2010+


houses['was_renovated'] = (houses['yr_renovated'] > 0).astype(int)
````

````python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsRegressor
from sklearn.pipeline import Pipeline

import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsRegressor

from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from scipy.stats import pearsonr
import matplotlib.pyplot as plt

def regression_metrics(y_true, y_pred):
    """
    Compute regression metrics (rounded to 3 decimals) from array-like inputs.
    Supports lists, numpy arrays, and pandas Series.
    """

    # Convert to numpy arrays and flatten
    y_true = np.asarray(y_true).ravel()
    y_pred = np.asarray(y_pred).ravel()

    # Metrics
    r2 = r2_score(y_true, y_pred)

    # Pearson can fail if one array is constant
    try:
        pearson, _ = pearsonr(y_true, y_pred)
    except Exception:
        pearson = np.nan

    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(y_true, y_pred)

    metrics_dict = {
        "r2": round(r2, 3),
        "pearson": round(float(pearson), 3) if pearson is not np.nan else np.nan,
        "mse": round(mse, 3),
        "rmse": round(rmse, 3),
        "mae": round(mae, 3),
    }

    title_str = (
        f"R²={metrics_dict['r2']} | "
        f"Pearson={metrics_dict['pearson']} | "
        f"RMSE={metrics_dict['rmse']} | "
        f"MSE={metrics_dict['mse']} | "
        f"MAE={metrics_dict['mae']}"
    )

    return metrics_dict, title_str


def plot_regression_scatter(y_true, y_pred, title_str=None):
    """
    Create a scatter plot of y_true vs y_pred with:
    - Ideal diagonal line
    - Confidence interval band (±1 std of residuals)
    - Scatter points
    - Labels and optional title
    """

    y_true = np.asarray(y_true).ravel()
    y_pred = np.asarray(y_pred).ravel()

    # Compute confidence band from residuals
    residuals = y_pred - y_true
    std = np.std(residuals)

    # Line for ideal diagonal
    xmin, xmax = min(y_true.min(), y_pred.min()), max(y_true.max(), y_pred.max())
    line = np.linspace(xmin, xmax, 200)

    plt.figure(figsize=(6, 6))

    # Scatter points
    plt.scatter(y_true, y_pred, alpha=0.6)

    # Ideal prediction diagonal
    plt.plot(line, line, linestyle="--", linewidth=2)

    # Confidence interval band
    plt.fill_between(
        line,
        line - std,
        line + std,
        alpha=0.2
    )

    # Axis labels
    plt.xlabel("True Values", fontsize=12)
    plt.ylabel("Predicted Values", fontsize=12)

    # Title
    if title_str:
        plt.title(title_str, fontsize=12)

    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()

def split_dataframe(df, target_col, test_size=0.2, random_state=42):
  """
  Splits a DataFrame into train and test sets.

  Parameters:
      df (pd.DataFrame): The input dataframe.
      target_col (str): Name of the target column.
      test_size (float): Fraction of data to use for test set (default 0.2).
      random_state (int): Random seed for reproducibility.

  Returns:
      pd.DataFrame, pd.DataFrame: train_df, test_df
  """
  train_df, test_df = train_test_split(
      df,
      test_size=test_size,
      random_state=random_state,
      stratify=df[target_col] if df[target_col].dtype == 'object' or df[target_col].dtype.name=='category' else None
  )
  return train_df, test_df

def train_predict_knn(train_df, test_df, target_column, n_neighbors=5):
  """
  Train a KNN regressor and predict on test data.
  Parameters:
  - train_df: pandas DataFrame containing training data (features + target)
  - test_df: pandas DataFrame containing test data (features only or with target)
  - target_column: string, name of the target column in train_df
  - n_neighbors: int, number of neighbors for KNN

  Returns:
  - model: trained KNeighborsRegressor
  - y_pred: predicted values for test_df
  """
  # Split features and target
  X_train = train_df.drop(columns=[target_column])
  y_train = train_df[target_column]
  X_test = test_df.drop(columns=[target_column], errors='ignore')

  # Initialize and train model
  model = KNeighborsRegressor(n_neighbors=n_neighbors)
  model.fit(X_train, y_train)

  # Predict
  y_pred = model.predict(X_test)

  return model, y_pred

def train_knn(X, y, n_neighbors=5, test_size=0.2, random_state=None):
  """
  Train a KNN regressor with scaling and return the trained model and predictions.

  Parameters:
  - X: Feature matrix
  - y: Target vector
  - n_neighbors: Number of neighbors for KNN
  - test_size: Fraction of data to use as test set
  - random_state: Random seed for reproducibility

  Returns:
  - knn_pipeline: Trained pipeline
  - y_pred: Predictions on the test set
  - X_test, y_test: Optional, test data for further evaluation
  """
  X_train, X_test, y_train, y_test = train_test_split(
      X, y, test_size=test_size, random_state=random_state
  )

  knn_pipeline = Pipeline([
      ("scaler", StandardScaler()),
      ("knn", KNeighborsRegressor(n_neighbors=n_neighbors))
  ])

  knn_pipeline.fit(X_train, y_train)
  y_pred = knn_pipeline.predict(X_test)

  return knn_pipeline, y_pred, X_test, y_test
````

````python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor

def train_rf(X, y, n_estimators=200, max_depth=None, min_samples_split=2,
min_samples_leaf=1, test_size=0.2, random_state=None):
  """
  Train a Random Forest regressor with optional scaling and return the trained model and predictions.

  ```
  Parameters:
  - X: Feature matrix
  - y: Target vector
  - n_estimators: Number of trees in the forest
  - max_depth: Maximum depth of each tree
  - min_samples_split: Minimum samples required to split an internal node
  - min_samples_leaf: Minimum samples required to be at a leaf node
  - test_size: Fraction of data to use as test set
  - random_state: Random seed for reproducibility

  Returns:
  - rf_pipeline: Trained pipeline
  - y_pred: Predictions on the test set
  - X_test, y_test: Test data for evaluation
  """
  X_train, X_test, y_train, y_test = train_test_split(
      X, y, test_size=test_size, random_state=random_state
  )

  rf_pipeline = Pipeline([
      ("scaler", StandardScaler()),
      ("rf", RandomForestRegressor(
          n_estimators=n_estimators,
          max_depth=max_depth,
          min_samples_split=min_samples_split,
          min_samples_leaf=min_samples_leaf,
          random_state=random_state,
          n_jobs=-1
      ))
  ])

  rf_pipeline.fit(X_train, y_train)
  y_pred = rf_pipeline.predict(X_test)

  return rf_pipeline, y_pred, X_test, y_test

rf_model, rf_y_pred, X_test, y_test = train_rf(X, y)
metrics_dic, metrics_title = regression_metrics(y_test, rf_y_pred)
plot_regression_scatter(y_test, rf_y_pred,metrics_title)
````

The KNN regression model shows stronger performance compared to the linear regression. With an R² of 0.756, it explains about 75.6% of the variation in house prices, which is an improvement over the linear model’s 70%. The RMSE is €191,894, lower than the linear model’s €212,017, meaning predictions are closer to actual values. This suggests KNN is better at capturing complex, nonlinear patterns in the data. Further improvements may be possible by tuning the number of neighbors, scaling features, or applying weighted distances.
